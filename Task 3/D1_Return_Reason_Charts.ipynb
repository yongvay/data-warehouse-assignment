# Student D — D1 Return Reason Analysis by Item Category

Run `d1_chart_exports.sql` first. This notebook uses the two CSV files generated from the Oracle warehouse; it contains no fabricated result data.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

CSV_DIR = Path('task3_csv')
OUT_DIR = Path('charts')
OUT_DIR.mkdir(exist_ok=True)
```

## Chart 1 — Return reason mix per item category

```python
df = pd.read_csv(CSV_DIR / 'd1_reason_mix.csv')
reason_cols = ['Missing', 'Broken', 'Expired', 'Wrong Item']
for c in reason_cols:
    df[c] = pd.to_numeric(df[c], errors='coerce').fillna(0)

ax = df.set_index('Category')[reason_cols].plot(
    kind='bar', stacked=True, figsize=(11, 6)
)
ax.set_title('Return Reason Mix by Item Category')
ax.set_xlabel('Item Category')
ax.set_ylabel('Quantity Returned')
ax.legend(title='Return Reason')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig(OUT_DIR / 'D1_Chart1_Return_Reason_Mix.png', dpi=220, bbox_inches='tight')
plt.show()
```

## Chart 2 — Refund value by reason category

`Fulfilment` represents Missing + Wrong Item. `Product Quality` represents Broken + Expired.

```python
pie = pd.read_csv(CSV_DIR / 'd1_refund_by_reason_category.csv')
pie['Refund Amount'] = pd.to_numeric(pie['Refund Amount'], errors='coerce').fillna(0)
pie = pie[pie['Refund Amount'] > 0]

fig, ax = plt.subplots(figsize=(7, 7))
ax.pie(
    pie['Refund Amount'],
    labels=pie['Reason Category'],
    autopct='%1.1f%%',
    startangle=90
)
ax.set_title('Refund Value by Return Reason Category')
plt.tight_layout()
plt.savefig(OUT_DIR / 'D1_Chart2_Refund_By_Reason_Category.png', dpi=220, bbox_inches='tight')
plt.show()
```
